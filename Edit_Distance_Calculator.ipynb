This code calculates editi distance between two sequences of the same length

```python
def edit_distance(s1, s2):
    if len(s1) != len(s2):
        raise ValueError("Sequences must be the same length")
# Calculate # of mismatches
    distance=0
    for a, b in zip(s1, s2):
        if a != b:
            distance += 1
# Return value
    return distance
# Run function with parameters 

#Create sequences that you want to find distance for
seq_a = "ATGCC"
seq_b = "ACGCT"   
edit_distance(seq_a, seq_b)
```
